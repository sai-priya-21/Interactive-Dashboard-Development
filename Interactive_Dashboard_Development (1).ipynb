```python
from dash import Dash, dcc, html
import plotly.express as px
import pandas as pd

# Sample data
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 17000, 21000, 25000],
    "Profit": [3000, 4000, 4500, 4200, 5000, 6000],
    "Customer_Satisfaction": [78, 82, 85, 80, 88, 90]
}
df = pd.DataFrame(data)

# Charts
fig_sales = px.line(df, x="Month", y="Sales", title="Monthly Sales Trend", markers=True)
fig_profit = px.bar(df, x="Month", y="Profit", title="Monthly Profit Comparison", color="Profit")
fig_scatter = px.scatter(df, x="Profit", y="Customer_Satisfaction",
                         title="Customer Satisfaction vs Profit",
                         size="Profit", color="Customer_Satisfaction", hover_name="Month")

# Dashboard layout
app = Dash(__name__)
app.layout = html.Div([
    html.H1("Business Performance Dashboard", style={"textAlign": "center"}),
    dcc.Graph(figure=fig_sales),
    dcc.Graph(figure=fig_profit),
    dcc.Graph(figure=fig_scatter)
])

# ✅ Updated run method
if __name__ == "__main__":
    app.run(debug=True)
```
